Cuaderno de lectura de la medición `adult/` del repositorio [ManPlaNet-datos](https://github.com/mmunozpl/ManPlaNet-datos). Respalda el artículo [mirar-ya-es-tratar](https://manpla.net/posts/mirar-ya-es-tratar/). Carga el fichero de al lado —o lo descarga del repositorio si se ejecuta fuera de él—, muestra la ficha de procedencia y dibuja una figura con matplotlib a secas. Solo lee; no vuelve a tomar la instantánea: para eso está `generar.py`.

*Reading notebook for this measurement: loads the file next to it, prints the provenance record and draws one figure. Column names are in Spanish; `GLOSARIO.md` gives the English form.*

```python
import io, json, urllib.request
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAW = "https://raw.githubusercontent.com/mmunozpl/ManPlaNet-datos/main/adult/"

def leer(nombre, **kw):
    """el fichero de al lado si existe; si no, el del repositorio."""
    p = Path(nombre)
    if p.exists():
        return pd.read_csv(p, **kw)
    return pd.read_csv(RAW + nombre, **kw)

def texto(nombre):
    p = Path(nombre)
    if p.exists():
        return p.read_text(encoding="utf-8")
    with urllib.request.urlopen(RAW + nombre, timeout=30) as r:
        return r.read().decode("utf-8")
```

## Ficha de procedencia

```python
print(texto("adult.names"))
```

## El dato

```python
cols = ["age","workclass","fnlwgt","education","education-num","marital-status","occupation","relationship","race","sex","capital-gain","capital-loss","hours-per-week","native-country","income"]
a = leer("adult.data.gz", header=None, names=cols, skipinitialspace=True, na_values="?")
print(len(a), "filas de entrenamiento ·", a.isna().any(axis=1).sum(), "con algún valor ausente")
a.head(15)
```

## Una figura

```python
fig, (x, y) = plt.subplots(1, 2, figsize=(11, 4))
a.income.value_counts(normalize=True).plot.bar(ax=x, rot=0); x.set_title("etiqueta"); x.set_ylabel("proporción")
a.isna().mean().sort_values(ascending=False).head(6).plot.bar(ax=y, rot=30); y.set_title("proporción de ausentes, seis columnas con más")
plt.tight_layout()
```
