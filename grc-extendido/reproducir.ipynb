# Competencias de GRC más allá del SGSI, medidas sobre datos reales

Cuaderno de lectura de la medición `grc-extendido/` del repositorio [ManPlaNet-datos](https://github.com/mmunozpl/ManPlaNet-datos): roles GRC del marco NICE, requisitos por categoría del ENS y de SP 800-53B, incidentes de VERIS con su curva de pérdidas, muestreo de pruebas de control, vigencia de los certificados del ENS y artículos de seguridad en las resoluciones de la AEPD. Carga los ficheros de al lado si existen y, si no, los del repositorio.

```python
import json, urllib.request
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAW = "https://raw.githubusercontent.com/mmunozpl/ManPlaNet-datos/main/grc-extendido/"

def leer(nombre, **kw):
    """el fichero de al lado si existe; si no, el del repositorio."""
    p = Path(nombre)
    if p.exists():
        return pd.read_csv(p, **kw)
    return pd.read_csv(RAW + nombre, **kw)

def texto(nombre):
    p = Path(nombre)
    if p.exists():
        return p.read_text(encoding="utf-8")
    with urllib.request.urlopen(RAW + nombre, timeout=30) as r:
        return r.read().decode("utf-8")

pd.set_option("display.width", 200)
```

## Ficha de procedencia

```python
print(texto("INSTANTANEA.md"))
```

## Roles GRC del marco NICE

```python
print(leer("nice_roles.csv").sort_values("rol", kind="stable").to_string(index=False))
e = leer("nice_enunciados.csv")
print("núcleo (4 roles o más):", (e.roles >= 4).sum(), "· de un solo rol:", (e.roles == 1).sum())
```

## Incidentes de VERIS y curva de pérdidas

```python
print(leer("veris_reparto.csv").to_string(index=False))
r = json.loads(texto("resumen.json"))["veris"]
print({k: r[k] for k in ("importes_usd", "usd_mediana", "lognormal_mu", "lognormal_sigma", "ks_p_corregido", "p90", "p95", "p99", "p95_ic", "p99_ic")})
c = leer("perdida_excedencia.csv")
ax = c.plot(x="perdida_usd", y="prob_superar", logx=True, legend=False, figsize=(8, 3.5))
ax.set_ylabel("probabilidad de superar"); plt.show()
print(leer("perdida_sensibilidad.csv").to_string(index=False))
```
