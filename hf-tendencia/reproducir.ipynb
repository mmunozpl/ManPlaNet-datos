Solo lee; no vuelve a tomar la instantánea: para eso está `generar.py`.

*Reading notebook for this measurement: loads the file next to it, prints the provenance record and draws one figure. Column names are in Spanish; `GLOSARIO.md` gives the English form.*

```python
import io, json, urllib.request
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAW = "https://raw.githubusercontent.com/mmunozpl/ManPlaNet-datos/main/hf-tendencia/"

def leer(nombre, **kw):
    """el fichero de al lado si existe; si no, el del repositorio."""
    p = Path(nombre)
    if p.exists():
        return pd.read_csv(p, **kw)
    return pd.read_csv(RAW + nombre, **kw)

def texto(nombre):
    p = Path(nombre)
    if p.exists():
        return p.read_text(encoding="utf-8")
    with urllib.request.urlopen(RAW + nombre, timeout=30) as r:
        return r.read().decode("utf-8")
```

## Ficha de procedencia

```python
print(texto("INSTANTANEA.md"))
```

## El dato

```python
t = leer("tendencia.csv")
print((t.descargas == 0).sum(), "con cero descargas ·", (t.con_fichero_consulta == 0).sum(), "sin fichero de consulta")
t.head(15)
```

## Una figura

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
con = t[t.descargas > 0]; sin = t[t.descargas == 0]
ax.scatter(con.edad_dias, con.descargas, s=18, label="con descargas")
ax.scatter(sin.edad_dias, [1]*len(sin), s=30, marker="x", color="C3", label="cero descargas (dibujadas en 1)")
ax.set_yscale("log"); ax.set_xlabel("edad del repositorio (días)"); ax.set_ylabel("descargas"); ax.legend(); ax.set_title("tendencia frente a descargas")
```
