# Las nueve amenazas de ISO/IEC 27090 frente a lo que ya se cataloga

Cuaderno de lectura de la medición `iso-27090/` del repositorio [ManPlaNet-datos](https://github.com/mmunozpl/ManPlaNet-datos). Las 170 técnicas y los 57 casos de estudio de MITRE ATLAS repartidos entre las nueve amenazas de la cláusula 6 de ISO/IEC 27090:2026, el anexo B y lo que queda fuera, con las correspondencias de NIST AI 100-2 E2025, OWASP Top 10 para LLM (2025) y el artículo 15.5 del Reglamento de IA. Carga los ficheros de al lado (o los descarga del repositorio si se ejecuta fuera de él), muestra la ficha de procedencia y dibuja una figura con matplotlib a secas. Solo lee: volver a tomar el dato es trabajo de `generar.py`.


```python
import urllib.request
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAW = "https://raw.githubusercontent.com/mmunozpl/ManPlaNet-datos/main/iso-27090/"

def leer(nombre, **kw):
    """el fichero de al lado si existe; si no, el del repositorio."""
    p = Path(nombre)
    if p.exists():
        return pd.read_csv(p, **kw)
    return pd.read_csv(RAW + nombre, **kw)

def texto(nombre):
    p = Path(nombre)
    if p.exists():
        return p.read_text(encoding="utf-8")
    with urllib.request.urlopen(RAW + nombre) as r:
        return r.read().decode("utf-8")
```

## Ficha de procedencia


```python
print(texto("INSTANTANEA.md"))
```

## El dato


```python
res = leer("resumen_27090.csv")
print(res[["clausula", "amenaza", "tecnicas", "realized", "casos", "incidentes"]].to_string(index=False))
nueve = res[res.clausula.str.match(r"^6\.")]
print("técnicas en las nueve amenazas:", int(nueve.tecnicas.sum()), "de", int(res.tecnicas.sum()))
casos = leer("atlas_casos_27090.csv")
print("casos que tocan alguna de las nueve:", int((casos.solo_fuera == 0).sum()), "de", len(casos))
```

## Una figura


```python
t = res[res.clausula != "fuera"].set_index("amenaza")[["tecnicas", "casos"]]
ax = t.plot.barh(figsize=(9, 5), width=0.8)
ax.invert_yaxis()
ax.set_xlabel("número")
ax.set_ylabel("")
ax.set_title("Técnicas y casos de estudio de MITRE ATLAS por amenaza de ISO/IEC 27090")
ax.legend(["técnicas", "casos de estudio"], frameon=False)
plt.tight_layout()
plt.show()
```
