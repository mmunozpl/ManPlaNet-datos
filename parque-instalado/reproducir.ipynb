Cuaderno de lectura de la medición `parque-instalado/` del repositorio [ManPlaNet-datos](https://github.com/mmunozpl/ManPlaNet-datos). Carga el fichero de al lado —o lo descarga del repositorio si se ejecuta fuera de él—, muestra la ficha de procedencia y dibuja una figura con matplotlib a secas. Solo lee; no vuelve a tomar la instantánea: para eso está `generar.py`.

*Reading notebook for this measurement: loads the file next to it, prints the provenance record and draws one figure. Column names are in Spanish; `GLOSARIO.md` gives the English form.*

```python
import io, json, urllib.request
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAW = "https://raw.githubusercontent.com/mmunozpl/ManPlaNet-datos/main/parque-instalado/"

def leer(nombre, **kw):
    """el fichero de al lado si existe; si no, el del repositorio."""
    p = Path(nombre)
    if p.exists():
        return pd.read_csv(p, **kw)
    return pd.read_csv(RAW + nombre, **kw)

def texto(nombre):
    p = Path(nombre)
    if p.exists():
        return p.read_text(encoding="utf-8")
    with urllib.request.urlopen(RAW + nombre, timeout=30) as r:
        return r.read().decode("utf-8")
```

## Ficha de procedencia

```python
print(texto("INSTANTANEA.md"))
```

## El dato

```python
e = leer("edades-por-anio.csv"); w = leer("windows-versiones.csv")
print(w[w.mes == w.mes.max()][["region", "mes", "win11", "win10", "win7"]].to_string(index=False))
e
```

## Una figura

```python
fig, (x, y) = plt.subplots(1, 2, figsize=(12, 4.2))
b = e.set_index("anio_alta")[["edad_0", "edad_1_4", "edad_5_9", "edad_10_mas"]]
b.plot.bar(stacked=True, ax=x, rot=0); x.set_title("altas del catálogo por edad del fallo (años)"); x.set_ylabel("altas")
m = w[w.region == "mundo"].set_index("mes")[["win11", "win10", "win7"]].astype(float)
m.plot(ax=y); y.set_title("Windows en escritorio, mundo (% StatCounter)"); y.tick_params(axis="x", rotation=45); plt.tight_layout()
```
