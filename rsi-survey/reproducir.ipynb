Carga los ficheros de al lado —o los descarga del repositorio—, muestra la ficha de procedencia y dibuja una figura con matplotlib a secas. Solo lee; `generar.py` vuelve a escribir la transcripción.

*Reading notebook for this measurement: loads the files next to it (or from the repository), prints the provenance note and draws one plain matplotlib figure. It only reads; `generar.py` rewrites the transcription.*

```python
import urllib.request
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAW = "https://raw.githubusercontent.com/mmunozpl/ManPlaNet-datos/main/rsi-survey/"

def leer(nombre, **kw):
    """el fichero de al lado si existe; si no, el del repositorio."""
    p = Path(nombre)
    return pd.read_csv(p, **kw) if p.exists() else pd.read_csv(RAW + nombre, **kw)

def texto(nombre):
    p = Path(nombre)
    if p.exists():
        return p.read_text(encoding="utf-8")
    with urllib.request.urlopen(RAW + nombre, timeout=30) as r:
        return r.read().decode("utf-8")
```

## La ecuación 4: el 78 % del margen, en todos los dominios

```python
h = leer("hci-eq4.csv").sort_values("hci_2026")
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.barh(h.dominio, h.hci_2026, color="C0", label="HCI 2026")
ax.barh(h.dominio, h.ganancia_ilustrativa, left=h.hci_2026, color="C1", alpha=0.5, label="ganancia de la ecuación 4")
ax.set_xlim(0, 100); ax.set_xlabel("índice de margen cerrado (HCI)"); ax.legend(); plt.tight_layout()
```
